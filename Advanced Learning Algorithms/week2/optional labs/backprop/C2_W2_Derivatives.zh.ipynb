## 导数的非正式定义

导数的正式定义可能因涉及极限和"趋于零"的值而让人望而生畏。其思想其实要简单得多。

一个函数的导数描述了当输入变量发生微小变化时，该函数的输出如何变化。

让我们以代价函数 $J(w)$ 为例。代价 $J$ 是输出，$w$ 是输入变量。
让我们给"微小变化"起个名字，叫 *epsilon* 或 $\epsilon$。我们使用这些希腊字母，因为在数学中传统上用 *epsilon*（$\epsilon$）或 *delta*（$\Delta$）来表示一个很小的值。你可以把它想成代表 0.001 或其他某个很小的值。

$$
\begin{equation}
\text{if } w \uparrow \epsilon \text{ causes }J(w) \uparrow \text{by }k \times \epsilon \text{ then}  \\
\frac{\partial J(w)}{\partial w} = k \tag{1}
\end{equation}
$$

这只是说，如果你把函数 $J(w)$ 的输入改变一点点，而输出改变了这一点点的 $k$ 倍，那么 $J(w)$ 的导数就等于 $k$。

让我们来试一试。让我们看看函数 $J(w) = w^2$ 在点 $w=3$、$\epsilon = 0.001$ 处的导数

````python
J = (3)**2
J_epsilon = (3 + 0.001)**2
k = (J_epsilon - J)/0.001    # difference divided by epsilon
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k:0.6f} ")
````

我们把输入值增大了一点点（0.001），导致输出从 9 变为 9.006001，增大了输入增量的 6 倍。参照上面的 (1)，这表示 $k=6$，所以 $\frac{\partial J(w)}{\partial w} \approx 6$。如果你熟悉微积分，你会知道用符号写出来是 $\frac{\partial J(w)}{\partial w} = 2 w$。当 $w=3$ 时它就是 6。我们上面的计算并不恰好等于 6，因为要完全正确的话，$\epsilon$ 需要是[无穷小的](https://www.dictionary.com/browse/infinitesimally)，或者说非常非常小。这就是为什么我们使用符号 $\approx$ 或 ~= 而不是 =。让我们看看如果把 $\epsilon$ 变小会发生什么。

````python
J = (3)**2
J_epsilon = (3 + 0.000000001)**2
k = (J_epsilon - J)/0.000000001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

随着我们减小 $\epsilon$ 的大小，该值越来越接近恰好为 6。你可以随意尝试进一步减小这个值。

## 求符号导数
在反向传播中，了解简单函数在任意输入值处的导数是有用的。换句话说，我们想知道的是"符号"导数，而不是"算术"导数。符号导数的一个例子是 $\frac{\partial J(w)}{\partial w} = 2 w$，即上面 $J(w) = w^2$ 的导数。有了符号导数，你就能求出该导数在任意输入值 $w$ 处的值。

如果你上过微积分课，你会熟悉数学家们为在给定表达式时求解导数而发展出的许多[求导法则](https://en.wikipedia.org/wiki/Differentiation_rules#Power_laws,_polynomials,_quotients,_and_reciprocals)。事实证明，这个过程已经被符号微分程序自动化了。Python 中的一个例子就是 [SymPy](https://www.sympy.org/en/index.html) 库。让我们来看看如何使用它。

### $J = w^2$
定义 python 变量及其符号名称。

````python
J, w = symbols('J, w')
````

定义并打印该表达式。请注意 SymPy 会生成一个 [latex](https://en.wikibooks.org/wiki/LaTeX/Mathematics) 字符串，它能生成一个美观易读的方程。

````python
J=w**2
J
````

使用 SymPy 的 `diff` 对 $J$ 关于 $w$ 的表达式求导。
请注意结果与我们前面的例子相符。

````python
dJ_dw = diff(J,w)
dJ_dw
````

通过用数值"替换"符号值，来在几个点上计算该导数的值。在第一个例子中，$w$ 被替换为 $2$。

````python
dJ_dw.subs([(w,2)])    # derivative at the point w = 2
````

````python
dJ_dw.subs([(w,3)])    # derivative at the point w = 3
````

````python
dJ_dw.subs([(w,-3)])    # derivative at the point w = -3
````

## $J = 2w$

````python
w, J = symbols('w, J')
````

````python
J = 2 * w
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,-3)])    # derivative at the point w = -3
````

把它与算术计算作比较

````python
J = 2*3
J_epsilon = 2*(3 + 0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

对于函数 $J=2w$，很容易看出：无论 $w$ 的起始值是多少，$w$ 的任何变化都会导致输出 $J$ 变化该量的 2 倍。我们的 NumPy 和算术结果都证实了这一点。

## $J = w^3$

````python
J, w = symbols('J, w')
````

````python
J=w**3
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,2)])   # derivative at the point w=2
````

把它与算术计算作比较

````python
J = (2)**3
J_epsilon = (2+0.001)**3
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````

## $J = \frac{1}{w}$

````python
J, w = symbols('J, w')
````

````python
J= 1/w
J
````

````python
dJ_dw = diff(J,w)
dJ_dw
````

````python
dJ_dw.subs([(w,2)])
````

把它与算术计算作比较

````python
J = 1/2
J_epsilon = 1/(2+0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
````
